# AI Job Market: Grouped Salary Analysis

This notebook compares salaries across experience levels, job
categories, countries, and work arrangements.

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../ai_jobs_market_2025_2026.csv")
df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
df.head()
```

## 1. Salary by experience level

We compare job count and salary statistics across experience levels.

```python
salary_by_experience = (
    df.groupby("experience_level")["annual_salary_usd"]
      .agg(["count", "mean", "median", "min", "max"])
      .sort_values("mean", ascending=False)
      .round(0)
)

salary_by_experience
```

```python
lead_mean = salary_by_experience.loc["Lead (10+ yrs)", "mean"]
entry_mean = salary_by_experience.loc["Entry (0-2 yrs)", "mean"]

salary_gap = lead_mean - entry_mean
growth_rate = salary_gap / entry_mean

print(f"Salary gap: ${salary_gap:,.0f}")
print(f"Lead premium over entry: {growth_rate:.1%}")
```

### Observation

Average salary rises consistently with experience. Lead-level roles
average $240,055, compared with $150,039 for entry-level roles, a
difference of about $90,016 or 60.0%.

The group sizes range from 364 to 385, so no experience group is
severely underrepresented in this dataset.

## 2. Salary by job category

We compare job counts and salary statistics across job categories.

```python
salary_by_category = (
    df.groupby("job_category")["annual_salary_usd"]
      .agg(["count", "mean", "median"])
      .sort_values("mean", ascending=False)
      .round(0)
)

salary_by_category
```

### Observation

Architecture has the highest average salary at $251,577, followed by
AI Engineering at $207,982. However, Architecture contains only 52
records, while AI Engineering contains 736, so the Architecture
estimate is based on a much smaller sample.

Business has the lowest average salary at $134,145. In most
categories, the mean is higher than the median, suggesting that
higher-paying roles pull the averages upward.

## 3. Salary by country

We compare salary levels and sample sizes across countries.

```python
salary_by_country = (
    df.groupby("country")["annual_salary_usd"]
      .agg(["count", "mean", "median"])
      .sort_values("mean", ascending=False)
      .round(0)
)

salary_by_country
```

## 4. Salary by country and experience level

We use a multi-index to compare salary within both country and
experience-level groups.

```python
salary_by_country_experience = (
    df.groupby(
        ["country", "experience_level"]
    )["annual_salary_usd"]
      .agg(["count", "mean", "median"])
      .round(0)
)

salary_by_country_experience
```

```python
salary_by_country_experience.loc[
    ["USA", "China", "India"]
]
```

```python
country_experience_table = (
    salary_by_country_experience
    .reset_index()
)

country_experience_table.head()
```

```python
selected_countries = country_experience_table[
    country_experience_table["country"].isin(
        ["USA", "China", "India"]
    )
]

selected_countries = selected_countries.sort_values(
    ["experience_level", "mean"],
    ascending=[True, False]
)

selected_countries
```

## 5. Salary by experience level and work arrangement

We compare work arrangements within each experience level to reduce
the effect of differences in seniority.

```python
salary_by_experience_and_work = (
    df.groupby(
        ["experience_level", "remote_work"]
    )["annual_salary_usd"]
      .agg(["count", "mean", "median"])
      .round(0)
)

salary_by_experience_and_work
```

```python
salary_by_experience_and_work["mean"].unstack()
```

### Observation

No single work arrangement has the highest average salary at every
experience level. On-site roles lead for entry and lead positions,
while fully remote roles lead for mid and senior positions.

For senior roles, fully remote and on-site salaries are almost
identical. Therefore, this dataset does not support the general claim
that remote work always pays more. Job category, country, and other
factors may also influence salary.

## Conclusion

- Average salary increases consistently with experience. Lead roles
  earn about 60% more than entry-level roles in this dataset.
- Architecture has the highest average salary, but its sample is much
  smaller than AI Engineering, so the result should be interpreted
  cautiously.
- The USA has the highest country-level average salary, although the
  comparison does not adjust for cost of living or differences in job
  composition.
- Fully remote work is not consistently the highest-paid arrangement
  after comparing roles within the same experience level.
