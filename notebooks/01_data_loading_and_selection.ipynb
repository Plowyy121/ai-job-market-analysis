```python
# TODO 1: Show the first 10 jobs located in the USA.
usa_jobs = df.loc[df["country"]=='USA', ["job_title", "annual_salary_usd", "city"]]
usa_jobs.head(10)
```

```python
# TODO 2: Find jobs with annual salaries above 200,000 USD.
high_salary_jobs = df.loc[df['annual_salary_usd']>200000, ["job_title", "experience_level", "annual_salary_usd"]]
high_salary_jobs.head(10)
```

```python
# TODO 3: Find remote LLM roles. Hint: combine two conditions with &.
remote_llm_jobs = df.loc[(df['is_remote_friendly']==1 )& (df['is_llm_role']==1), ["job_title", "country", "remote_work", "annual_salary_usd"]]
remote_llm_jobs.head(10)
```
